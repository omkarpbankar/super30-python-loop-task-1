# Python For-Loop Fundamentals

This notebook practices iteration with `for`, `range()`, strings, lists, tuples, sets, and dictionaries.

## 1. Print numbers from 1 to 100

```python
for number in range(1, 101):
    print(number)
```

## 2. Print even numbers from 1 to 100

```python
for number in range(1, 101):
    if number % 2 == 0:
        print(number)
```

## 3. Print odd numbers from 1 to 100

```python
for number in range(1, 101):
    if number % 2 != 0:
        print(number)
```

## 4. Multiplication table from 1 to 20

```python
n = int(input("Enter an integer: "))

for multiplier in range(1, 21):
    print(f"{n} x {multiplier} = {n * multiplier}")
```

## 5. Sum numbers from 1 to n

```python
n = int(input("Enter a positive integer: "))
total = 0

for number in range(1, n + 1):
    total += number

print(f"The sum from 1 to {n} is {total}.")
```

## 6. Factorial without a built-in factorial function

```python
number = int(input("Enter a non-negative integer: "))
factorial = 1

# Multiply the running result by every integer from 1 through number.
for value in range(1, number + 1):
    factorial *= value

print(f"{number}! = {factorial}")
```

## 7. Numbers divisible by 3

```python
numbers = [12, 7, 9, 20, 33, 42, 8, 15]

for number in numbers:
    if number % 3 == 0:
        print(number)
```

## 8. Languages and their lengths

```python
languages = ["Python", "Java", "C++", "JavaScript", "Go"]

for language in languages:
    print(f"{language}: {len(language)} characters")
```

## 9. Iterate through a dictionary

```python
student = {
    "name": "Rahul",
    "age": 22,
    "course": "Data Science",
    "city": "Bangalore"
}

for key, value in student.items():
    print(f"{key}: {value}")
```

## 10. Count vowels in a user-provided string

```python
user_text = input("Enter a string (press Enter for an example): " )
if not user_text:
    user_text = "Iteration makes Python powerful"

vowel_count = 0
for character in user_text.lower():
    if character in "aeiou":
        vowel_count += 1

print(f"The string contains {vowel_count} vowels.")
```

## 11. Reverse a string with a for loop

```python
text = input("Enter a string to reverse (press Enter for an example): " )
if not text:
    text = "Python"

reversed_text = ""
# Add each new character before the current result.
for character in text:
    reversed_text = character + reversed_text

print(f"Reversed string: {reversed_text}")
```

## 12. Find the largest number without `max()`

```python
values = [18, 4, 27, 11, 35, 9]
largest = values[0]

for value in values[1:]:
    if value > largest:
        largest = value

print(f"The largest number is {largest}.")
```
